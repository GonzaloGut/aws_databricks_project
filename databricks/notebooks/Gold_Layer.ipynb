```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.format("delta")\
            .load("s3://aws-s3-bucket-airflow/silver/obt")

display(df)
```

## Bookings Per City

```python
df_bookings = df.groupBy("city").agg(count("booking_id").alias("total_bookings")).sort("total_bookings",ascending=False)

df_bookings.write.format("delta")\
        .mode("overwrite")\
        .saveAsTable("airflow_gold.gold.bookingsCity")
```

## Airports Per Country

```python
df_airports = df.groupBy("country").agg(count("airport_id").alias("total_airports")).sort("total_airports",ascending=False)

df_airports.write.format("delta")\
        .mode("overwrite")\
        .saveAsTable("airflow_gold.gold.airportsCountry")
```
